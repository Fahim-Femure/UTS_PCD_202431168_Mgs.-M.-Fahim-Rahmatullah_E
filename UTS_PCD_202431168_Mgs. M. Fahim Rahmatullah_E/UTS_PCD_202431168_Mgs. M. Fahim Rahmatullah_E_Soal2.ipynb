```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
```

```python
IMAGE_PATH = 'foto_backlight_3.jpeg'

img_bgr = cv2.imread(IMAGE_PATH)
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(7, 5))
plt.imshow(img_rgb)
plt.title('Gambar Asli', fontsize=14)
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(7, 5))
plt.imshow(img_gray, cmap='gray')
plt.title('Gambar Gray', fontsize=14)
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
BETA = 80
img_bright = np.clip(img_gray.astype(np.int32) + BETA, 0, 255).astype(np.uint8)

plt.figure(figsize=(7, 5))
plt.imshow(img_bright, cmap='gray')
plt.title('Gambar Gray yang Dipercerah', fontsize=14)
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
ALPHA_CONTRAST = 1.2
img_contrast = cv2.convertScaleAbs(img_gray, alpha=ALPHA_CONTRAST, beta=0)

plt.figure(figsize=(7, 5))
plt.imshow(img_contrast, cmap='gray')
plt.title('Gambar Gray yang Diperkontras', fontsize=14)
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
ALPHA_FINAL = 1.4
BETA_FINAL = 70
img_bright_contrast = cv2.convertScaleAbs(img_gray, alpha=ALPHA_FINAL, beta=BETA_FINAL)

plt.figure(figsize=(7, 5))
plt.imshow(img_bright_contrast, cmap='gray')
plt.title('Gambar Gray yang Dipercerah dan Diperkontras', fontsize=14)
plt.axis('off')
plt.tight_layout()
plt.show()
```
